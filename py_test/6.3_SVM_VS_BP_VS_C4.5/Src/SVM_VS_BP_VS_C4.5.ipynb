```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split

# 加载数据
data = pd.read_csv('../Data/krkopt.data', header=None).values
# 特征数值化,将a-h转化为1-8
for i in [0,2,4]:
    data[:,i] = np.where(data[:,i]=='a', 1, data[:,i])
    data[:,i] = np.where(data[:,i]=='b', 2, data[:,i])
    data[:,i] = np.where(data[:,i]=='c', 3, data[:,i])
    data[:,i] = np.where(data[:,i]=='d', 4, data[:,i])
    data[:,i] = np.where(data[:,i]=='e', 5, data[:,i])
    data[:,i] = np.where(data[:,i]=='f', 6, data[:,i])
    data[:,i] = np.where(data[:,i]=='g', 7, data[:,i])
    data[:,i] = np.where(data[:,i]=='h', 8, data[:,i])
# 标签数值化,'draw'视为-1,其余视为1
data[:,-1] = np.where(data[:,-1]=='draw', -1, 1)

# 数据归一化
for i in range(6):
    data[:,i] = (data[:,i]-np.mean(data[:,i]))/np.std(data[:,i])

# 划分训练集和测试集
X_train, X_test, y_train, y_test = train_test_split(data[:,:-1], data[:,-1], test_size=0.3, random_state=42)
```
